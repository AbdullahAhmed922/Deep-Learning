```python
import numpy as np
import keras
```

```python
xs = np.array([-1, 0, 1, 2, 3, 4])
ys = np.array([-3, -1, 1, 3, 5, 7])
```

```python
model = keras.Sequential()
model.add(layer = keras.Input(shape=(1,)))
model.add(layer = keras.layers.Dense(units=1))
```

```python
model.summary()
```

```python
model.compile(
    optimizer=keras.optimizers.SGD(learning_rate=0.1),
    loss=keras.losses.MeanSquaredError(),
    metrics=[
        keras.metrics.MeanAbsoluteError(),
    ],
)
```

```python
model.fit(x=xs, y=ys, epochs=500, verbose=1)
```

```python
y_pred = model.predict(x=np.array([10]))
```

```python
y_pred
```

```python
# Q2: Getting the final linear regression equation
weights, bias = model.get_weights()
m = weights[0][0]
b = bias[0]

print(f"Coefficient (Weight): {m:.4f}")
print(f"Intercept (Bias): {b:.4f}")
print(f"Final Equation: y = {m:.4f}x + ({b:.4f})")
```

```python
# Q1: Making a prediction for 12 years of experience
y_pred_12 = model.predict(x=np.array([12]))
print(f"Prediction for 12 years of experience: {y_pred_12[0][0]}")
```
