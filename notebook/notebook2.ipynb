```python
import numpy as np
import pandas as pd
import keras
```

```python
# The dataset is located in the 'data' folder relative to the notebook
df = pd.read_csv("./data/salary_dataset.csv")
print(df.head())    
print(df.columns)    
```

```python
X = df[["YearsExperience"]]
y = df["Salary"]
```

```python
model = keras.Sequential()
model.add(layer = keras.Input(shape=(1,)))
model.add(layer = keras.layers.Dense(units=1))
```

```python
model.summary()
```

```python
model.compile(
    optimizer=keras.optimizers.SGD(learning_rate=0.0001),
    loss=keras.losses.MeanSquaredError(),
    metrics=[
        keras.metrics.MeanAbsoluteError(),
    ],
)
```

```python
model.fit(x=X, y=y, epochs=500, verbose=1)
```

```python
y_pred = model.predict(x=np.array(object=[20]))
```

```python
y_pred
```

```python
# Q2: Getting the final linear regression equation
weights, bias = model.get_weights()
m = weights[0][0]
b = bias[0]

print(f"Coefficient (Weight): {m:.4f}")
print(f"Intercept (Bias): {b:.4f}")
print(f"Final Equation: y = {m:.4f}x + ({b:.4f})")
```

```python
# Q1: Making a prediction for 12 years of experience
y_pred_12 = model.predict(x=np.array([20]))
print(f"Prediction for 12 years of experience: {y_pred_12[0][0]}")
```
